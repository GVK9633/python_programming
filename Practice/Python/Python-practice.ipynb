````python
# Examples of Relational Operators 
a = 13
b = 33
   
# a > b is False 
print(a > b) 
   
# a < b is True 
print(a < b) 
   
# a == b is False 
print(a == b) 
   
# a != b is True 
print(a != b) 
   
# a >= b is False 
print(a >= b) 
   
# a <= b is True 
print(a <= b)
````

````python
# Examples of Bitwise operators 
a = 10
b = 4
   
# Print bitwise AND operation   
print(a & b) 
   
# Print bitwise OR operation 
print(a | b) 
   
# Print bitwise NOT operation  
print(~a) 
   
# print bitwise XOR operation  
print(a ^ b) 
   
# print bitwise right shift operation  
print(a >> 2) 
   
# print bitwise left shift operation  
print(a << 2)
````

Here’s a **line-by-line explanation** of your code that demonstrates the usage of **Bitwise Operators** in Python:

---

### 🔢 **Variable Assignment**

```python
a = 10
b = 4
```

* `a = 10` → binary: `1010`
* `b = 4`  → binary: `0100`

---

### 1️⃣ **Bitwise AND (`&`)**

```python
print(a & b)
```

* Binary: `1010 & 0100`
* Result: `0000` → `0` (only bits that are **1 in both** remain 1)

✅ **Output:** `0`

---

### 2️⃣ **Bitwise OR (`|`)**

```python
print(a | b)
```

* Binary: `1010 | 0100`
* Result: `1110` → `14` (bits that are **1 in either**)

✅ **Output:** `14`

---

### 3️⃣ **Bitwise NOT (`~`)**

```python
print(~a)
```

* Binary of `a` = `0000 1010`
* `~a` flips all bits → becomes a negative number in Python

> Bitwise NOT of `a = 10`:
> \~10 = `-11`
> (Uses **2’s complement** internally)

✅ **Output:** `-11`

---

### 4️⃣ **Bitwise XOR (`^`)**

```python
print(a ^ b)
```

* Binary: `1010 ^ 0100`
* Result: `1110` → `14` (bits are 1 **if different** in a and b)

✅ **Output:** `14`

---

### 5️⃣ **Bitwise Right Shift (`>>`)**

```python
print(a >> 2)
```

* Binary: `1010` → shift right by 2 bits → `0010`
* This is integer division by `2² = 4`

✅ **Output:** `2`

---

### 6️⃣ **Bitwise Left Shift (`<<`)**

```python
print(a << 2)
```

* Binary: `1010` → shift left by 2 bits → `101000`
* This multiplies the number by `2² = 4`

✅ **Output:** `40`

---

## 🧠 Summary of Bitwise Operators:

| Operator | Meaning            | Example   | Result |     |    |
| -------- | ------------------ | --------- | ------ | --- | -- |
| `&`      | AND                | `10 & 4`  | 0      |     |    |
| \`       | \`                 | OR        | \`10   | 4\` | 14 |
| `~`      | NOT (invert bits)  | `~10`     | -11    |     |    |
| `^`      | XOR (exclusive OR) | `10 ^ 4`  | 14     |     |    |
| `>>`     | Right Shift        | `10 >> 2` | 2      |     |    |
| `<<`     | Left Shift         | `10 << 2` | 40     |     |    |
