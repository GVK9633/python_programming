## ✅ **NumPy Interview Cheat Sheet** (Beginner to Experienced)

---

### 🟢 **1. NumPy Basics**

| Concept         | Command/Function     | Example               |
| --------------- | -------------------- | --------------------- |
| Import NumPy    | `import numpy as np` | –                     |
| Create array    | `np.array()`         | `np.array([1, 2, 3])` |
| Shape of array  | `.shape`             | `arr.shape`           |
| Size of array   | `.size`              | `arr.size`            |
| Data type       | `.dtype`             | `arr.dtype`           |
| Change type     | `.astype()`          | `arr.astype(float)`   |
| Array dimension | `.ndim`              | `arr.ndim`            |

---

### 🟡 **2. Creating Arrays**

| Task             | Function              | Example                                |
| ---------------- | --------------------- | -------------------------------------- |
| Zeros            | `np.zeros(shape)`     | `np.zeros((2,3))`                      |
| Ones             | `np.ones(shape)`      | `np.ones((3,3))`                       |
| Constant value   | `np.full()`           | `np.full((2,2), 7)`                    |
| Identity matrix  | `np.eye()`            | `np.eye(3)`                            |
| Range of numbers | `np.arange()`         | `np.arange(1, 10, 2)`                  |
| Even spacing     | `np.linspace()`       | `np.linspace(0, 1, 5)`                 |
| Random values    | `np.random.rand()`    | `np.random.rand(3,2)`                  |
| Random int       | `np.random.randint()` | `np.random.randint(1, 10, size=(2,2))` |

---

### 🟡 **3. Array Operations**

| Operation        | Syntax            | Example        |
| ---------------- | ----------------- | -------------- |
| Element-wise add | `+`               | `a + b`        |
| Subtraction      | `-`               | `a - b`        |
| Multiplication   | `*`               | `a * b`        |
| Division         | `/`               | `a / b`        |
| Power            | `**`              | `a ** 2`       |
| Dot product      | `np.dot()` or `@` | `np.dot(a, b)` |
| Transpose        | `.T`              | `a.T`          |

---

### 🟡 **4. Indexing & Slicing**

| Task             | Syntax                 | Example      |
| ---------------- | ---------------------- | ------------ |
| Index            | `arr[i]`               | `arr[2]`     |
| Slice            | `arr[start:stop:step]` | `arr[1:5:2]` |
| 2D Index         | `arr[i, j]`            | `arr[0,1]`   |
| Fancy Indexing   | `arr[[0,2]]`           | –            |
| Boolean Indexing | `arr[arr > 3]`         | –            |

### 🟠 **5. Useful NumPy Functions**

| Function      | Purpose            | Example               |
| ------------- | ------------------ | --------------------- |
| `np.sum()`    | Sum                | `np.sum(arr, axis=1)` |

| `np.median()` | Median             | –                     |
| `np.std()`    | Standard deviation | –                     |
| `np.var()`    | Variance           | –                     |
| `np.min()`    | Minimum            | –                     |
| `np.max()`    | Maximum            | –                     |
| `np.argmin()` | Index of min       | –                     |
| `np.argmax()` | Index of max       | –                     |

### 🔵 **6. Reshaping & Combining**

| Task               | Function           | Example                 |
| ------------------ | ------------------ | ----------------------- |
| Reshape            | `arr.reshape()`    | `arr.reshape(3,2)`      |
| Flatten            | `arr.flatten()`    | –                       |
| Stack vertically   | `np.vstack()`      | `np.vstack((a,b))`      |
| Stack horizontally | `np.hstack()`      | `np.hstack((a,b))`      |
| Split arrays       | `np.split()`       | `np.split(arr, 2)`      |

### 🔁 **9. Broadcasting Rules**

* Smaller dimensions are automatically expanded.
* Example:

```python
a = np.array([[1,2,3], [4,5,6]])
b = np.array([1, 2, 3])
print(a + b)
```

### 🧪 **10. Common Interview Patterns**

| Question              | Concept                                        |
| --------------------- | ---------------------------------------------- |
| Remove duplicates     | `np.unique(arr)`                               |
| Find most frequent    | `np.bincount(arr).argmax()`                    |
| Reverse array         | `arr[::-1]`                                    |
| Normalize data        | `(arr - arr.min()) / (arr.max() - arr.min())`  |
| Moving average        | `np.convolve(arr, np.ones(3)/3, mode='valid')` |
| Replace NaN with mean | Use `np.nanmean()` and `np.isnan()`            |
