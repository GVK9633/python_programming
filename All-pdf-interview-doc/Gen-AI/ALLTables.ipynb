# 📌 **Prompt Engineering Cheat Sheet for GenAI Interviews**

| **Technique**                            | **Definition**                                                                                    | **Example Prompt**                                                                                                                           | **When to Use**                                                                                                  |
| ---------------------------------------- | ------------------------------------------------------------------------------------------------- | -------------------------------------------------------------------------------------------------------------------------------------------- | ---------------------------------------------------------------------------------------------------------------- |
| **Zero-Shot Prompting**                  | Asking the model to perform a task **without giving any examples**.                               | *"Summarize this paragraph in 3 bullet points."*                                                                                             | ✅ Well-known tasks with clear instructions.<br>✅ Quick answers without context.                                  |
| **One-Shot Prompting**                   | Providing **one example** of the task before asking for the final output.                         | *"Example: 'I am happy' → 'Je suis heureux'.<br>Translate: 'They are friends' → "*                                                           | ✅ Simple custom formats.<br>✅ Helps model understand expected output style.                                      |
| **Few-Shot Prompting**                   | Giving **2–5 examples** to set a pattern or context for the model.                                | *"Translate English to French:<br>1. I am happy → Je suis heureux.<br>2. She is a teacher → Elle est professeur.<br>3. They are friends → "* | ✅ Complex or uncommon tasks.<br>✅ When zero-shot results are poor.<br>✅ Helps enforce specific format/structure. |
| **Chain-of-Thought (CoT)**               | Instructing the model to **think step-by-step** before answering.                                 | *"Solve this problem step-by-step:<br>If a pen costs \$1 and a book costs \$3, how much for 4 pens and 2 books?"*                            | ✅ Math, logic, reasoning problems.<br>✅ Reduces hallucinations in multi-step reasoning.                          |
| **Self-Consistency with CoT**            | Asking the model to **generate multiple reasoning paths and choose the most likely answer**.      | *"List 3 possible solutions to this puzzle, then select the most logical one."*                                                              | ✅ Complex reasoning where one answer may be wrong if model takes shortcut thinking.                              |
| **RAG (Retrieval-Augmented Generation)** | Combining **external knowledge sources (databases, documents)** with the LLM to improve accuracy. | *"Using the knowledge base on Hugging Face datasets, answer: 'What is the ESC-50 dataset used for?'"*                                        | ✅ When information is **not in the model’s training data**.<br>✅ For enterprise knowledge-based chatbots.        |
| **Instruction + Context Prompting**      | Providing **specific instructions and context** to guide responses.                               | *"Act as a data science interviewer. Ask me 5 technical ML questions with increasing difficulty."*                                           | ✅ To set a persona or role for the model.<br>✅ Useful in chatbots, simulations, and interview practice.          |
| **Few-Shot + CoT Hybrid**                | Giving examples **and** asking for step-by-step reasoning.                                        | *"Example: Solve 2+3=5 (Step-by-step: 2+3=5). Now, solve 15+27 step-by-step."*                                                               | ✅ Tasks needing both structure (examples) and logical reasoning.                                                 |

---

## ✅ **Key Tips for Prompt Engineering**

* Use **clear instructions** (what, how, format, tone).
* Use **delimiters** like `"""` or `###` to separate instructions from input text.
* Be **specific about output format** (bullet points, JSON, table).
* Use **few-shot examples** for uncommon tasks.
* Use **CoT or RAG** for complex reasoning or when model lacks context.

Unlike discriminative artificial intelligence that performs classification tasks, modern Generative Artificial Intelligence uses machine learning and deep neural networks to understand and conditionally generate new examples from complex data distributions. What are some well-established architectures used to develop Generative AI?


✅ **Correct Answers:**

* **Embeddings to represent high-dimensional, complex data**
* **Variational Autoencoders (VAE)**
* **Generative Adversarial Networks (GAN)**
* **Diffusion Models**
* **Transformers for Large Language Models (LLMs)**
* **Neural Radiance Fields (NeRF)**

---

### **Explanation of Each Option**

1️⃣ **✅ Embeddings:**

* **Purpose:** Transform complex data (text, image, audio) into **high-dimensional numeric vectors** that capture semantic meaning.
* **Use case:** Fundamental for most Generative AI models, enabling similarity search, semantic understanding, and contextual generation.

---

2️⃣ **✅ Variational Autoencoders (VAE):**

* **Architecture:** Encoder → Latent Space → Decoder.
* **Purpose:** Learn compressed representation of data and **generate new samples** from latent space.
* **Use case:** Image, video, and speech generation.

---

3️⃣ **✅ Generative Adversarial Networks (GAN):**

* **Architecture:** Generator (creates data) + Discriminator (judges real vs fake).
* **Purpose:** Adversarial training to create **realistic synthetic data**.
* **Use case:** Face generation, deepfakes, video synthesis.

---

4️⃣ **✅ Diffusion Models:**

* **Architecture:** Gradually add noise to data and then **reverse the process** to generate new, highly detailed samples.
* **Use case:** Image generation (e.g., **DALL·E 2, Stable Diffusion**).

---

5️⃣ **✅ Transformers (LLMs):**

* **Architecture:** Attention-based neural networks for processing **sequential data**.
* **Purpose:** Generate human-like text by predicting the next token.
* **Use case:** **GPT, LaMBDA, LLaMA, Bard**, code generation, conversational AI.

---

6️⃣ **✅ Neural Radiance Fields (NeRF):**

* **Purpose:** Convert **2D image datasets into 3D scenes** by modeling light radiance in a 3D space.
* **Use case:** 3D content generation for VR, AR, gaming, and simulations.

---

All the listed architectures (**1, 2, 3, 4, 5, 6**) are **well-established components of modern Generative AI**.
