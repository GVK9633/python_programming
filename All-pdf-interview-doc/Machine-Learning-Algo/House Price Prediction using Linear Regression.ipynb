---

### 🏠 **Project: House Price Prediction using Linear Regression**

#### ✅ Use Case:

Predicting the **price of a house** given its **area (in square feet)**.

---

### 📦 Step 1: Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

---

### 🧾 Step 2: Create Dataset

```python
# Area in square feet
area = np.array([500, 750, 1000, 1250, 1500, 1750, 2000]).reshape(-1, 1)

# Corresponding price in thousands
price = np.array([100, 150, 200, 250, 300, 350, 400])
```

---

### 🤖 Step 3: Train the Linear Regression Model

```python
model = LinearRegression()
model.fit(area, price)

# Print model coefficients
print("Slope (m):", model.coef_[0])
print("Intercept (c):", model.intercept_)
```

---

### 📈 Step 4: Predict New Price

```python
new_area = np.array([[1800]])
predicted_price = model.predict(new_area)
print(f"Predicted price for 1800 sq.ft: {predicted_price[0]:.2f}K")
```

---

### 📊 Step 5: Plot the Regression Line

```python
plt.scatter(area, price, color='blue', label='Actual Data')
plt.plot(area, model.predict(area), color='red', label='Regression Line')
plt.xlabel('Area (sq.ft)')
plt.ylabel('Price (in 1000s)')
plt.title('House Price Prediction')
plt.legend()
plt.grid(True)
plt.show()
```

---

### 📌 Summary:

* **Algorithm Used**: Linear Regression
* **Formula**: `Price = m * Area + c`
* **Libraries**: NumPy, Pandas, Matplotlib, scikit-learn

````python
 # Step 1: Import Libraries
 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
````

````python
### 🧾 Step 2: Create Dataset

# Area in square feet
area = np.array([500, 750, 1000, 1250, 1500, 1750, 2000]).reshape(-1, 1)

# Corresponding price in thousands
price = np.array([100, 150, 200, 250, 300, 350, 400])

````

````python
### 🤖 Step 3: Train the Linear Regression Model

model = LinearRegression()
model.fit(area, price)

# Print model coefficients
print("Slope (m):", model.coef_[0])
print("Intercept (c):", model.intercept_)
````

````python
### 📈 Step 4: Predict New Price


new_area = np.array([[1800]])
predicted_price = model.predict(new_area)
print(f"Predicted price for 1800 sq.ft: {predicted_price[0]:.2f}K")
````

````python
### 📊 Step 5: Plot the Regression Line
plt.figure(figsize=(10, 6))
plt.scatter(area, price, color='blue', label='Actual Data')
plt.plot(area, model.predict(area), color='red', label='Regression Line')
plt.xlabel('Area (sq.ft)')
plt.ylabel('Price (in 1000s)')
plt.title('House Price Prediction')
plt.legend()
plt.grid(True)
plt.show()
````

Great question! To **know whether a Linear Regression model predicts the correct price**, you evaluate it using **performance metrics** and **visualizations**. Here's how you can **verify the accuracy** of your model:

---

## ✅ 1. **Split the Data (Train/Test Split)**

You should split your dataset into:

* **Training data**: to train the model
* **Test data**: to evaluate how well the model generalizes

```python
from sklearn.model_selection import train_test_split

# Split data into 80% train and 20% test
X_train, X_test, y_train, y_test = train_test_split(area, price, test_size=0.2, random_state=42)
```

---

## ✅ 2. **Train the Model on Training Data**

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

---

## ✅ 3. **Make Predictions on Test Data**

```python
y_pred = model.predict(X_test)
```

---

## ✅ 4. **Use Evaluation Metrics**

Use these metrics to check model performance:

### 📌 a. Mean Absolute Error (MAE)

Average of absolute prediction errors.

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)
print(f"Mean Absolute Error: {mae:.2f}")
```

### 📌 b. Mean Squared Error (MSE) / RMSE

Penalizes larger errors more than MAE.

```python
from sklearn.metrics import mean_squared_error
import numpy as np

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
print(f"Root Mean Squared Error: {rmse:.2f}")
```

### 📌 c. R² Score (Coefficient of Determination)

* `R² = 1` → perfect model
* `R² = 0` → predicts average
* `R² < 0` → worse than average

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)
print(f"R² Score: {r2:.2f}")
```

---

## ✅ 5. **Visualize Predictions vs Actuals**

```python
plt.scatter(X_test, y_test, color='blue', label='Actual Price')
plt.scatter(X_test, y_pred, color='red', label='Predicted Price')
plt.plot(area, model.predict(area), color='green', label='Regression Line')
plt.xlabel('Area (sq.ft)')
plt.ylabel('Price (in 1000s)')
plt.title('Prediction vs Actual')
plt.legend()
plt.grid(True)
plt.show()
```

---

## 🧠 How to Know It's Good?

| Metric   | Ideal Value          | Interpretation      |
| -------- | -------------------- | ------------------- |
| MAE      | Close to 0           | Small average error |
| RMSE     | Close to 0           | Smaller is better   |
| R² Score | Close to 1 (max = 1) | Higher = better fit |
