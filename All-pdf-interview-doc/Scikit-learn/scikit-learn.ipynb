**Q4. Load a built-in dataset (Iris).**

```python
from sklearn.datasets import load_iris
data = load_iris()
print(data.feature_names)
```

**Q5. How to split data into training and testing sets?**

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2)
```

A: Predicting categorical outcomes (e.g., spam or not spam).

**Q9. Train a logistic regression model.**

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train, y_train)
```

**Q10. Predict and calculate accuracy.**

```python
from sklearn.metrics import accuracy_score
y_pred = model.predict(X_test)
accuracy_score(y_test, y_pred)
```

**Q11. Confusion matrix?**

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, y_pred)
```

**Q12. Print classification report.**

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

A: When a model performs well on training data but poorly on unseen data.

**Q14. What is underfitting?**
A: When a model is too simple to capture the patterns in the data.
