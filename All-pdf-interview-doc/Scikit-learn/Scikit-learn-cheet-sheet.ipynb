## ✅ **Scikit-learn Interview Cheat Sheet (Beginner to Expert)**

---

### 🟢 **1. Basic Setup**

| Task           | Code                                                                                                                                 |
| -------------- | ------------------------------------------------------------------------------------------------------------------------------------ |
| Import library | `import sklearn`                                                                                                                     |
| Check version  | `sklearn.__version__`                                                                                                                |
| Load dataset   | `from sklearn.datasets import load_iris`<br>`data = load_iris()`                                                                     |
| Split data     | `from sklearn.model_selection import train_test_split`<br>`X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)` |

---

### 🟡 **2. Preprocessing**

| Task               | Function                            | Example                             |
| ------------------ | ----------------------------------- | ----------------------------------- |
| Standardize        | `StandardScaler()`                  | `StandardScaler().fit_transform(X)` |
| Normalize          | `MinMaxScaler()`                    | –                                   |
| Encode categorical | `OneHotEncoder()`, `LabelEncoder()` | –                                   |
| Handle missing     | `SimpleImputer()`                   | –                                   |
| Pipeline           | `Pipeline()`                        | Combine preprocessing + model       |

```python
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])
```

### 🔴 **6. Model Evaluation**

#### 🔹 **Classification Metrics**

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix
```

| Metric                | Usage                                   |
| --------------------- | --------------------------------------- |
| Accuracy              | `accuracy_score(y_test, y_pred)`        |
| Precision             | `precision_score(y_test, y_pred)`       |
| Recall                | `recall_score(y_test, y_pred)`          |
| F1 Score              | `f1_score(y_test, y_pred)`              |
| Confusion Matrix      | `confusion_matrix(y_test, y_pred)`      |
| Classification Report | `classification_report(y_test, y_pred)` |

#### 🔹 **Regression Metrics**

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

| Metric   | Usage                                 |
| -------- | ------------------------------------- |
| MAE      | `mean_absolute_error(y_test, y_pred)` |
| MSE      | `mean_squared_error(y_test, y_pred)`  |
| RMSE     | `np.sqrt(mse)`                        |
| R2 Score | `r2_score(y_test, y_pred)`            |

### 🧪 **8. Pipeline & GridSearch**

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV

pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('model', RandomForestClassifier())
])

param_grid = {
    'model__n_estimators': [10, 50, 100],
    'model__max_depth': [None, 10, 20]
}

grid = GridSearchCV(pipe, param_grid, cv=5)
grid.fit(X_train, y_train)
```
