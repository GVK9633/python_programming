**Q61. What is multiprocessing in Python? How is it different from multithreading?**

* `multiprocessing` uses multiple processes with separate memory space.
* `multithreading` uses threads within the same process and affected by GIL.

**Q62. How do you use decorators with arguments?**

```python
def decorator(msg):
    def wrapper(fn):
        def inner(*args, **kwargs):
            print(msg)
            return fn(*args, **kwargs)
        return inner
    return wrapper

@decorator("Hello")
def greet():
    print("Welcome")
```

**Q63. What is a coroutine in Python?**
Coroutines allow suspension and resumption of functions using `async def`, `await`.

```python
import asyncio
async def main():
    await asyncio.sleep(1)
```

**Q64. Explain the usage of `functools.lru_cache`.**
Caches the results of expensive function calls.

```python
from functools import lru_cache
@lru_cache(maxsize=1000)
def fib(n):
    if n < 2: return n
    return fib(n-1) + fib(n-2)
```

**Q65. What are weak references?**

**Q73. How do you create a singleton class in Python?**

```python
class Singleton:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance
```
