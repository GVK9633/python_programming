## 🐼 Pandas Interview Questions & Cheatsheet

### 🔹 1. What is Pandas? Why use it?

**Answer:**
Pandas is a powerful open-source Python library used for data analysis and manipulation. It provides two main data structures:

* `Series` (1D)
* `DataFrame` (2D)

---

### 🔹 2. How do you create a DataFrame?

**Example:**

```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'Salary': [50000, 60000, 70000]
}

df = pd.DataFrame(data)
print(df)
```

---

### 🔹 3. How to read/write CSV files?

```python
df = pd.read_csv('data.csv')          # Reading
df.to_csv('output.csv', index=False)  # Writing
```

---

### 🔹 4. How to view top/bottom rows?

```python
df.head(5)   # First 5 rows
df.tail(5)   # Last 5 rows
```

---

### 🔹 5. How to get info and stats?

```python
df.info()
df.describe()
df.shape      # (rows, cols)
df.columns    # List of column names
```

---

### 🔹 6. How to filter rows?

```python
df[df['Age'] > 30]                 # Condition
df[df['Name'].str.contains('A')]  # String matching
```

---

### 🔹 7. How to add new column?

```python
df['Tax'] = df['Salary'] * 0.1
```

---

### 🔹 8. How to apply a function?

```python
df['Salary_after_tax'] = df['Salary'].apply(lambda x: x * 0.9)
```

---

### 🔹 9. How to sort data?

```python
df.sort_values('Age', ascending=True)
```

---

### 🔹 10. How to group data?

```python
df.groupby('Age')['Salary'].mean()
```

---

### 🔹 11. How to handle missing values?

```python
df.isnull().sum()
df.fillna(0)             # Replace nulls
df.dropna()              # Drop rows with nulls
```

---

### 🔹 12. How to merge/join DataFrames?

```python
pd.merge(df1, df2, on='ID', how='inner')
```

---

### 🔹 13. Difference between loc vs iloc?

```python
df.loc[0:2, ['Name', 'Age']]    # label-based
df.iloc[0:2, [0, 1]]            # index-based
```

---

### 🔹 14. How to pivot data?

```python
df.pivot_table(values='Salary', index='Age', columns='Department', aggfunc='mean')
```

---

### 🔹 15. How to reset and set index?

```python
df = df.set_index('Name')
df = df.reset_index()
```

---

### 🔹 16. Convert data types?

```python
df['Age'] = df['Age'].astype(float)
```

---

### 🔹 17. Remove duplicates?

```python
df.drop_duplicates()
```

---

### 🔹 18. Rename columns?

```python
df.rename(columns={'Age': 'Years'})
```

---

### 🔹 19. Iterate over rows?

```python
for index, row in df.iterrows():
    print(row['Name'], row['Salary'])
```

### 🔹 20. Export to Excel?
