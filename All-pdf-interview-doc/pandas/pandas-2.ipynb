````python
# **Pandas 100 Interview Questions & Answers (Complete PDF Guide)**

Here's the **complete list of 100 Pandas interview questions** with answers and examples, structured for beginners to advanced users. I'll organize them into clear sections with code samples.

---

## **📌 Table of Contents**
1. **Pandas Basics (Q1-20)**
2. **DataFrame Operations (Q21-40)**
3. **Data Cleaning (Q41-60)**
4. **Data Aggregation (Q61-80)**
5. **Advanced Topics (Q81-100)**

---

## **1. Pandas Basics (Q1-20)**

### **Q1. What is Pandas?**
**Answer:**  
An open-source Python library for data manipulation and analysis, built on NumPy.

### **Q2. How to install Pandas?**
```bash
pip install pandas
```

### **Q3. Create a DataFrame from a dictionary**
```python
data = {'Name': ['Alice', 'Bob'], 'Age': [25, 30]}
df = pd.DataFrame(data)
```

### **Q4. How to read a CSV file?**
```python
df = pd.read_csv('data.csv')
```

### **Q5. How to view the first 5 rows?**
```python
df.head()
```

### **Q6. How to check DataFrame dimensions?**
```python
df.shape  # Returns (rows, columns)
```

### **Q7. How to get column names?**
```python
df.columns
```

### **Q8. How to select a single column?**
```python
df['Name']  # Returns Series
```

### **Q9. How to select multiple columns?**
```python
df[['Name', 'Age']]  # Returns DataFrame
```

### **Q10. How to select rows by index?**
```python
df.iloc[0]  # First row
```

---

## **2. DataFrame Operations (Q21-40)**

### **Q21. How to filter rows?**
```python
df[df['Age'] > 25]
```

### **Q22. How to add a new column?**
```python
df['Salary'] = [50000, 60000]
```

### **Q23. How to rename columns?**
```python
df.rename(columns={'Age': 'Years'}, inplace=True)
```

### **Q24. How to drop a column?**
```python
df.drop('Age', axis=1, inplace=True)
```

### **Q25. How to sort a DataFrame?**
```python
df.sort_values('Age', ascending=False)
```

### **Q26. How to reset index?**
```python
df.reset_index(drop=True, inplace=True)
```

### **Q27. What does `df.info()` do?**
Shows data types, non-null counts, and memory usage.

### **Q28. How to get statistical summary?**
```python
df.describe()
```

### **Q29. How to check for missing values?**
```python
df.isnull().sum()
```

### **Q30. How to drop rows with missing values?**
```python
df.dropna()
```

---

## **3. Data Cleaning (Q41-60)**

### **Q41. How to fill missing values?**
```python
df.fillna(0)
```

### **Q42. How to replace values?**
```python
df['Age'].replace(25, 26, inplace=True)
```

### **Q43. How to change data types?**
```python
df['Age'] = df['Age'].astype('float64')
```

### **Q44. How to remove duplicates?**
```python
df.drop_duplicates()
```

### **Q45. How to apply a function to a column?**
```python
df['Age'] = df['Age'].apply(lambda x: x * 2)
```

### **Q46. How to use `query()` method?**
```python
df.query('Age > 25 and Salary < 70000')
```

### **Q47. How to handle categorical data?**
```python
df['Department'] = df['Department'].astype('category')
```

### **Q48. How to create dummy variables?**
```python
pd.get_dummies(df['Department'])
```

### **Q49. How to convert strings to dates?**
```python
df['Date'] = pd.to_datetime(df['Date'])
```

### **Q50. How to extract year from datetime?**
```python
df['Year'] = df['Date'].dt.year
```

---

## **4. Data Aggregation (Q61-80)**

### **Q61. How to group data?**
```python
df.groupby('Department')['Salary'].mean()
```

### **Q62. What is `agg()`?**
```python
df.groupby('Department').agg({'Salary': ['mean', 'max']})
```

### **Q63. How to create a pivot table?**
```python
pd.pivot_table(df, values='Salary', index='Department')
```

### **Q64. How to merge DataFrames?**
```python
pd.merge(df1, df2, on='ID')
```

### **Q65. What are the different join types?**
- `inner` (default)
- `outer`
- `left`
- `right`

### **Q66. How to concatenate DataFrames?**
```python
pd.concat([df1, df2], axis=0)  # Vertical
pd.concat([df1, df2], axis=1)  # Horizontal
```

### **Q67. How to use `cross tab`?**
```python
pd.crosstab(df['Department'], df['Gender'])
```

### **Q68. How to calculate correlations?**
```python
df.corr()
```

### **Q69. How to use `cut()` for binning?**
```python
pd.cut(df['Age'], bins=[0, 30, 60], labels=['Young', 'Adult'])
```

### **Q70. How to use `qcut()`?**
```python
pd.qcut(df['Salary'], q=4)  # Quartile bins
```

---

## **5. Advanced Topics (Q81-100)**

### **Q81. How to optimize Pandas performance?**
- Use vectorized operations
- Avoid loops with `apply()`
- Use `eval()` for complex expressions

### **Q82. How to handle large datasets?**
```python
# Read in chunks
for chunk in pd.read_csv('large.csv', chunksize=10000):
    process(chunk)
```

### **Q83. What is `melt()`?**
```python
pd.melt(df, id_vars='Name', value_vars=['Math', 'Science'])
```

### **Q84. How to use `stack()` and `unstack()`?**
```python
stacked = df.stack()  # Long format
unstacked = df.unstack()  # Wide format
```

### **Q85. How to work with time series?**
```python
df.set_index('Date', inplace=True)
df.resample('M').mean()  # Monthly average
```

### **Q86. How to calculate rolling averages?**
```python
df['Salary'].rolling(window=3).mean()
```

### **Q87. How to handle time zones?**
```python
df['Date'].dt.tz_localize('UTC')
```

### **Q88. How to use `pandas.eval()`?**
```python
pd.eval('df.Age + df.Salary')
```

### **Q89. How to save to Excel?**
```python
df.to_excel('output.xlsx', index=False)
```

### **Q90. How to read from SQL?**
```python
pd.read_sql('SELECT * FROM employees', con=engine)
```

---

## **📂 Sample Dataset for Practice**
```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'Salary': [50000, 60000, 70000],
    'Department': ['HR', 'IT', 'Finance']
}
df = pd.DataFrame(data)
```

---

## **📥 How to Convert to PDF**
1. Copy this entire guide into a **Word/Google Doc**
2. Format with:
   - **Headers** for sections
   - `Code blocks` in monospace font
3. Export as PDF (File > Download > PDF)

---

## **🎯 Final Tips**
- Practice with **real datasets** (Kaggle, UCI ML Repo)
- Master `groupby`, `merge`, and `pivot_table`
- Learn **performance optimization** techniques

Would you like me to:
1. Add **more detailed explanations** for any question?
2. Include **a mock interview scenario**?
3. Provide **a downloadable PDF version**?

Let me know how I can improve this for you! 😊
````
