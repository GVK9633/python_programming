# **🔹 Beginner Level (1-30)**
### **1. What is Pandas?**
**Answer:**  
Pandas is an open-source Python library for **data manipulation and analysis**, built on NumPy. It provides **DataFrame** and **Series** structures for handling tabular data.

### **2. How to create a DataFrame?**
**Answer:**  
```python
import pandas as pd

# From a dictionary
data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age': [25, 30, 35]}
df = pd.DataFrame(data)
print(df)
```
**Output:**
```
      Name  Age
0    Alice   25
1      Bob   30
2  Charlie   35
```

### **3. How to read a CSV file?**
**Answer:**  
```python
df = pd.read_csv('data.csv')
```

### **4. How to select a column?**
**Answer:**  
```python
df['Name']  # Returns a Series
df[['Name']]  # Returns a DataFrame
```

### **5. How to filter rows?**
**Answer:**  
```python
df[df['Age'] > 30]  # People older than 30
```

### **6. What is `df.head()`?**
**Answer:**  
Shows the **first 5 rows** of the DataFrame.

### **7. How to get DataFrame summary?**
**Answer:**  
```python
df.info()  # Data types & non-null counts
df.describe()  # Statistical summary
```

### **8. How to rename columns?**
**Answer:**  
```python
df.rename(columns={'Age': 'Years'}, inplace=True)
```

### **9. How to drop a column?**
**Answer:**  
```python
df.drop('Age', axis=1, inplace=True)
```

### **10. How to handle missing values?**
**Answer:**  
```python
df.dropna()  # Drop rows with NaN
df.fillna(0)  # Fill NaN with 0
```

# **🔹 Intermediate Level (31-70)**
### **31. How to group data?**
**Answer:**  
```python
df.groupby('Department')['Salary'].mean()
```

### **32. What is `apply()`?**
**Answer:**  
Applies a function to each row/column:
```python
df['Age'].apply(lambda x: x + 1)  # Increment age by 1
```

### **33. How to sort a DataFrame?**
**Answer:**  
```python
df.sort_values('Age', ascending=False)
```

### **34. How to create a pivot table?**
**Answer:**  
```python
pd.pivot_table(df, values='Salary', index='Department', columns='Year')
```

### **35. How to merge DataFrames?**
**Answer:**  
```python
pd.merge(df1, df2, on='ID', how='inner')
```

### **36. What is `pd.concat()`?**
**Answer:**  
Combines DataFrames vertically/horizontally:
```python
pd.concat([df1, df2], axis=0)  # Vertical stack
```

### **37. How to handle duplicates?**
**Answer:**  
```python
df.drop_duplicates()  # Remove duplicates
```

### **38. How to change data types?**
**Answer:**  
```python
df['Age'] = df['Age'].astype('float64')
```

### **39. How to use `query()`?**
**Answer:**  
Filter using SQL-like syntax:
```python
df.query('Age > 30 and Salary < 50000')
```

### **40. How to reset index?**
**Answer:**  
```python
df.reset_index(drop=True, inplace=True)
```

# **🔹 Advanced Level (71-100)**
### **71. How to optimize Pandas performance?**
**Answer:**  
- Use `df.apply()` instead of loops  
- Convert object dtypes to `category`  
- Use `pd.eval()` for complex operations  

### **72. How to handle large datasets?**
**Answer:**  
- Use `chunksize` in `read_csv()`  
- Use `dask` for parallel processing  

### **73. What is `pd.cut()`?**
**Answer:**  
Bins continuous data into intervals:
```python
pd.cut(df['Age'], bins=[0, 30, 60, 100], labels=['Young', 'Adult', 'Senior'])
```

### **74. How to use `melt()`?**
**Answer:**  
Converts wide to long format:
```python
pd.melt(df, id_vars='Name', value_vars=['Math', 'Science'])
```

### **75. How to handle time series data?**
**Answer:**  
```python
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)
df.resample('M').mean()  # Monthly average
```
