### 71. What are NumPy strides?

**Answer:** Strides tell how many bytes to step in each dimension when traversing an array.

```python
arr = np.array([[1, 2], [3, 4]])
arr.strides
```

---

### 72. What is a view in NumPy?

**Answer:** A view is a new array object that looks at the same data.

```python
a = np.array([1, 2, 3])
b = a.view()
b[0] = 100
print(a)  # Also changes
```

---

### 73. How to detect if an array is a view?

```python
b.base is not None
```

### 74. How to use advanced indexing with arrays?

### 75. What is memory layout in NumPy (`C` vs `F`)?

* `'C'`: Row-major (default)
* `'F'`: Column-major

```python
np.array([[1, 2], [3, 4]], order='F')
```

### 76. What is vectorization and why is it faster?

**Answer:** Operating on whole arrays without explicit loops.

```python
arr ** 2  # Fast
```

### 87. How to replace NaNs with column means?

```python
arr = np.array([[1, np.nan, 3], [4, 5, np.nan]])
col_means = np.nanmean(arr, axis=0)
inds = np.where(np.isnan(arr))
arr[inds] = np.take(col_means, inds[1])
```

### 88. How to compare performance of loops vs vectorization?

```python
import time
a = np.random.rand(1000000)
start = time.time()
a ** 2
print("Vectorized:", time.time() - start)
```
