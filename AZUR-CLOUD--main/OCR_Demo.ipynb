```python
from azure.cognitiveservices.vision.computervision import ComputerVisionClient
from msrest.authentication import CognitiveServicesCredentials
import time

# Azure credentials
key = 'FlIxsbvC7iLaddMB6TwblmGCWmJjJ6MD2A5UAiSf7scEyne5tv1hJQQJ99BFACYeBjFXJ3w3AAAFACOGnUZW'
endpoint = "https://east-us-rg.cognitiveservices.azure.com/"

# Initialize client
credentials = CognitiveServicesCredentials(key)
client = ComputerVisionClient(endpoint, credentials)

# Local image path
filename = r"/Users/gvijaykumarachary/Desktop/MyComputer/E-Drive/DataScience/Repos/python_programming/AZUR-CLOUD--main/Driving-License.jpg"

# Open the local image file in binary mode and call read_in_stream
with open(filename, "rb") as image_stream:
    rawHttpResponse = client.read_in_stream(image_stream, language="en", raw=True)
```

```python
# Get operation ID to poll results
operation_location = rawHttpResponse.headers["Operation-Location"]
operation_id = operation_location.split("/")[-1]
```

```python
# Polling loop
while True:
    result = client.get_read_result(operation_id)
    if result.status.lower() not in ['notstarted', 'running']:
        break
    time.sleep(1)

# Print recognized text
if result.status.lower() == 'succeeded':
    for page in result.analyze_result.read_results:
        for line in page.lines:
            print(line.text)
```
