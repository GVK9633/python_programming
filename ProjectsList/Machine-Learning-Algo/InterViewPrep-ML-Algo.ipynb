## 🚀 1. MACHINE LEARNING ALGORITHMS OVERVIEW

### 🔵 A. **Supervised Learning**

Learn from labeled data (i.e., input and known output).

#### 1. **Linear Regression**

* **Use case**: Predicting a continuous value.
* **Example**: Predicting house prices based on area.
* **Formula**: `y = mx + c`

#### 2. **Logistic Regression**

* **Use case**: Binary classification (0/1).
* **Example**: Predict if an email is spam or not.
* **Output**: Probability → then thresholded to 0 or 1.

#### 3. **Decision Tree**

* **Use case**: Classification or regression using if-else rules.
* **Example**: Classify customers as “High/Medium/Low” credit risk.

#### 4. **Random Forest**

* **Use case**: Better accuracy than a single decision tree by using multiple trees (bagging).
* **Example**: Predict whether a customer will churn.

#### 5. **Support Vector Machine (SVM)**

* **Use case**: Classification with a margin of separation.
* **Example**: Handwriting recognition (digit classification).

#### 6. **K-Nearest Neighbors (KNN)**

* **Use case**: Classifies based on closest data points.
* **Example**: Recommending movies based on user similarity.

#### 7. **Naive Bayes**

* **Use case**: Classification using probabilities.
* **Example**: Spam detection in Gmail (fast & simple).

#### 8. **Gradient Boosting / XGBoost / LightGBM**

* **Use case**: Powerful models for structured/tabular data.
* **Example**: Predict loan default using multiple features.
* **Strength**: Works well with complex patterns and missing data.

---

### 🔴 B. **Unsupervised Learning**

No labeled output; goal is to discover patterns.

#### 1. **K-Means Clustering**

* **Use case**: Group similar data points.
* **Example**: Segment customers based on buying behavior.

#### 2. **Hierarchical Clustering**

* **Use case**: Tree-based grouping.
* **Example**: Document topic grouping.

#### 3. **PCA (Principal Component Analysis)**

* **Use case**: Dimensionality reduction.
* **Example**: Reduce 100 features to top 2 for visualization.

---

### 🟢 C. **Semi-Supervised Learning**

* Combines small labeled + large unlabeled data.
* **Example**: Google Photos face recognition (some tagged, rest inferred).

---

### 🟡 D. **Reinforcement Learning**

* Learns from actions and rewards (agent-environment loop).
* **Example**: AlphaGo, self-driving cars, trading bots.

## 🧠 Tips to Decide the Best Algorithm

1. **Data Size**:

   * Small: Naive Bayes, Logistic Regression
   * Large: XGBoost, Neural Networks

2. **Speed vs Accuracy**:

   * Speed: Logistic Regression, Naive Bayes
   * Accuracy: Random Forest, XGBoost

3. **Interpretability**:

   * High: Decision Tree, Linear Regression
   * Low: Neural Networks, XGBoost

4. **Data Type**:

   * Text: Naive Bayes, Logistic Regression
   * Image: CNN
   * Time Series: LSTM

## 🧠 Machine Learning Algorithms Cheat Sheet

### 🔹 1. **Linear Regression**

* **Use for**: Predicting a continuous numeric value (regression).
* **Example**: Predict house prices based on area and number of rooms.
* **Assumption**: Linear relationship between input and output.
* ✅ Simple, interpretable
* ❌ Fails with non-linearity, multicollinearity

### 🔹 2. **Logistic Regression**

* **Use for**: Binary classification (0/1, Yes/No).
* **Example**: Predict if an email is spam or not.
* ✅ Fast, interpretable
* ❌ Doesn’t capture complex patterns

---

### 🔹 3. **Decision Tree**

* **Use for**: Classification & Regression
* **Example**: Decide loan approval based on income, age, credit score.
* ✅ Easy to visualize
* ❌ Can overfit (prune or use ensemble)

### 🔹 4. **Random Forest**

* **Use for**: Classification/Regression with better generalization
* **Example**: Predict stock prices using multiple features.
* ✅ Robust, handles non-linearity, feature importance
* ❌ Slower, less interpretable

---

### 🔹 5. **Support Vector Machine (SVM)**

* **Use for**: Binary classification (works well with clear margins)
* **Example**: Classify tumor as malignant or benign.
* ✅ Effective in high dimensions
* ❌ Slow on large datasets

---

### 🔹 6. **K-Nearest Neighbors (KNN)**

* **Use for**: Classification (or regression with avg of neighbors)
* **Example**: Recognize handwritten digits (0-9)
* ✅ No training phase
* ❌ Slow on large datasets, sensitive to scaling

---

### 🔹 7. **Naive Bayes**

* **Use for**: Text classification (e.g., spam detection)
* **Example**: Classify news articles by topic.
* ✅ Very fast, good for NLP
* ❌ Assumes feature independence

---

### 🔹 8. **Gradient Boosting (GBM, XGBoost, LightGBM, CatBoost)**

* **Use for**: Highly accurate models for classification/regression
* **Example**: Kaggle winning solutions like fraud detection
* ✅ Handles missing data, categorical features
* ❌ Needs tuning, slow to train

---

### 🔹 9. **K-Means Clustering**

* **Use for**: Unsupervised clustering
* **Example**: Customer segmentation by purchase behavior
* ✅ Fast and simple
* ❌ Assumes spherical clusters, sensitive to k

---

### 🔹 10. **DBSCAN / Hierarchical Clustering**

* **Use for**: Clustering non-spherical data or density-based
* **Example**: Group spatial locations based on density
* ✅ No need to define k
* ❌ Not good with varying density

---

### 🔹 11. **Principal Component Analysis (PCA)**

* **Use for**: Dimensionality reduction
* **Example**: Reduce features from image data before modeling
* ✅ Speeds up training
* ❌ Loss of interpretability

---

## ✅ How to Choose the Right Algorithm?

| Criteria              | Go with…                                               |
| --------------------- | ------------------------------------------------------ |
| Small tabular data    | Logistic Regression, Decision Tree, SVM                |
| Large + complex data  | Random Forest, XGBoost, LightGBM                       |
| Fast prototype        | Naive Bayes, Logistic Regression, KNN                  |
| High interpretability | Linear/Logistic Regression, Decision Trees             |
| Image/Text data       | Deep Learning (CNN/RNN) or XGBoost for TF-IDF features |
| Unlabeled data        | K-Means, DBSCAN                                        |
| Imbalanced data       | Use ensemble with class weights / SMOTE + XGBoost      |
