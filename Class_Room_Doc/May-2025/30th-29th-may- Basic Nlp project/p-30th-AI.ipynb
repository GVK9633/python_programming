```python

from nltk.util import bigrams,trigrams,ngrams
# Bigrams
```

```python
string = "I am learning Natural Language Processing from 830am batch"
quotes_tokens = word_tokenize(string)
quotes_tokens
```

```python
quotes_bigran = list(bigrams(quotes_tokens))
print(quotes_bigran)
```

```python
from nltk.stem import PorterStemmer #filter root word from a word
# Example: running -> run, ran -> ran, easily -> easi, fairly -> fair
pst = PorterStemmer()
pst.stem("running"), pst.stem("ran"), pst.stem("easily"), pst.stem("fairly")
```

```python
pst.stem("happily"), pst.stem("happiness"), pst.stem("happy"), pst.stem("happier")
```

```python
words_to_stem = ["running", "ran", "easily", "fairly", "happily", "happiness", "happy", "happier"]
for word in words_to_stem:
    print(f"{word} -> {pst.stem(word)}")
```

```python
from nltk.stem import wordnet
from nltk.stem import WordNetLemmatizer
# Example: running -> run, ran -> ran, easily -> easily, fairly -> fair
wnl = WordNetLemmatizer()
wnl.lemmatize("running", pos='v'), wnl.lemmatize("ran", pos='v'), wnl.lemmatize("easily", pos='a'), wnl.lemmatize("fairly", pos='a')
wnl.lemmatize("happily", pos='a'), wnl.lemmatize("happiness", pos='n'), wnl.lemmatize("happy", pos='a'), wnl.lemmatize("happier", pos='a')
words_to_lemmatize = ["running", "ran", "easily", "fairly", "happily", "happiness", "happy", "happier"]
for word in words_to_lemmatize:
    print(f"{word} -> {wnl.lemmatize(word, pos='v')}")
    print(f"{word} -> {wnl.lemmatize(word, pos='a')}")
    print(f"{word} -> {wnl.lemmatize(word, pos='n')}")
```
