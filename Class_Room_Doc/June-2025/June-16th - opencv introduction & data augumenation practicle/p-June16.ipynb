```python
import tensorflow as tf
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import ImageDataGenerator, array_to_img, img_to_array, load_img
```

```python
datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.4,
    height_shift_range=0.4,
    shear_range=0.4,
    zoom_range=0.4,
    horizontal_flip=True,
    fill_mode='nearest',# try 'constant', 'reflect', 'wrap'
    rescale=1./255
)
```

```python
img = load_img('./puppy-1047521_1280.jpg')  # Load an image file
```

```python
img
```

```python
x= img_to_array(img)  # Convert the image to a numpy array
x = x.reshape((1,) + x.shape)  # Reshape the array to (1, height, width, channels)
i = 0
for batch in datagen.flow(x, batch_size=1):
    i += 1
    img = array_to_img(batch[0])  # Convert the batch back to an image
    img.save(f'./augmented_puppy_{i}.jpg')  # Save the augmented image
    if i > 5:  # Stop after generating 20 images
        break
print(f"Generated {i} augmented images.")
```
