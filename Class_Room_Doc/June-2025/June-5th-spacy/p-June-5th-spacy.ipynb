```python
import spacy
from spacy.lang.en.stop_words import STOP_WORDS
from string import punctuation
```

```python
stopwords = list(STOP_WORDS)
stopwords
```

```python
len(stopwords)
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp(newText)
doc
```

```python
tokens = [token.text for token in doc]
tokens
```

```python
#use case can you summarize the paragraph and suggest me the importance from the text
# how to find out the important words from the text
#based on the frequency of the words
# para = sentnece
# sentence= token
#token score = frequency of the word in the paragraph
```

```python
word_frequencies = {}
for word in doc:
    if word.text.lower() not in stopwords and word.text not in punctuation:
        if word.text not in word_frequencies.keys():
            word_frequencies[word.text] = 1
        else:
            word_frequencies[word.text] += 1
word_frequencies
```

```python
len(word_frequencies)
```

```python
max_frequency = max(word_frequencies.values())
max_frequency
```

```python
# waieghted frequency is 11 with all the words
## 1/11,2/11
for word in word_frequencies.keys():
    word_frequencies[word] = word_frequencies[word] / max_frequency
    
```

```python
word_frequencies
```

```python
sentence_tokens = [sent for sent in doc.sents]
sentence_tokens
```

```python
len(sentence_tokens)
```

```python
sentence_tokens
```

```python
sentence_scores = {}
for sent in sentence_tokens:
    for word in sent:
        if word.text.lower() in word_frequencies.keys():
            if sent not in sentence_scores.keys():
                sentence_scores[sent] = word_frequencies[word.text.lower()]
            else:
                sentence_scores[sent] += word_frequencies[word.text.lower()]
sentence_scores
```

```python
len(sentence_scores)
```

```python
# larest sentence and lower setnence
from heapq import nlargest
```

```python
#15
select_length = int(len(sentence_tokens) * 0.4)
select_length
```

```python
#sorting the sentence scores and selecting the top sentences
summary_sentences = nlargest(select_length, sentence_scores, key=sentence_scores.get)
summary_sentences
```

```python
finalsummary = ' '.join([word.text for word in summary_sentences])
print(finalsummary)
```
