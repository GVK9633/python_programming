```python
import numpy as np
import pandas as pd
import tensorflow as tf
```

```python
tf
```

```python
dataset = pd.read_csv('./Churn_Modelling.csv')
```

```python
dataset
```

```python
x= dataset.iloc[:, 3:-1].values
y = dataset.iloc[:, -1].values
```

```python
dataset.head()
```

```python
x
```

```python
y
```

```python
dataset.head(2)
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
le = LabelEncoder()
x[:, 2] = le.fit_transform(x[:, 2])
```

```python
x
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(
    transformers=[('encoder', OneHotEncoder(), [1])],
    remainder='passthrough'
)
x = np.array(ct.fit_transform(x))
```

```python
x
```

```python
from sklearn.preprocessing import StandardScaler
sc= StandardScaler()
x = sc.fit_transform(x)
x
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
print(x_train.shape, x_test.shape, y_train.shape, y_test.shape)
```

```python
#Lets build ANN modle building architecture
```

```python
ann = tf.keras.models.Sequential()# establish the layer one after another
```

```python
# 6 nurons created -input layer
ann.add(tf.keras.layers.Dense(units=6,))
```

```python
# Adding the activation function hiden layer-hidden layer-1
ann.add(tf.keras.layers.Dense(units=6, activation='relu'))
```

```python
ann.add(tf.keras.layers.Dense(units=5, activation='relu'))#layer-hidden layer-2
```

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu'))#layer-hidden layer-3
```

```python
#output layer
ann.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))
```

```python
ann.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Fitting the ANN to the Training set
ann.fit(x_train, y_train, batch_size=32, epochs=50)
```

```python
# Predicting the Test set results
y_pred = ann.predict(x_test)
y_pred = (y_pred > 0.5)
```

```python
# Making the Confusion Matrix
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
# Accuracy
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
print(f'Accuracy: {accuracy * 100:.2f}%')
```
