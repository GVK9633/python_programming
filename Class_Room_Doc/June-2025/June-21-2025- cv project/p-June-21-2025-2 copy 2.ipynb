```python
import cv2
import numpy as np

cap=cv2.VideoCapture(0)

while True:
    _,frame = cap.read()
    hsv_frmae = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    
    low_blue = np.array([94, 80, 2])
    high_blue = np.array([126, 255, 255])
    
    # low_red = np.array([0, 100, 100])
    # high_red = np.array([10, 255, 255])
    
    blue_mask = cv2.inRange(hsv_frmae, low_blue, high_blue)
    blue_result = cv2.bitwise_and(frame, frame, mask=blue_mask)
    # Display the red result
    
    cv2.imshow('frame', frame)
    cv2.imshow('Red Result', blue_result)
    cv2.imshow('Red Mask', blue_mask)

    key= cv2.waitKey(1)
    if key == ord('q'):
        break
  
```
