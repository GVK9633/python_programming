```python
import cv2
import numpy as np

cap=cv2.VideoCapture(0)

while True:
    _,frame = cap.read()
    hsv_frmae = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    
#     lower_green = np.array([40, 100, 100])
#     upper_green = np.array([80, 255, 255])
    
    # low_green = np.array([25, 52, 72])
    # high_green= np.array([102, 255, 255])
    
    low_green = np.array([40, 100, 100])
    high_green= np.array([102, 255, 255])

  
  
    green_mask = cv2.inRange(hsv_frmae, low_green, high_green)
    green_result = cv2.bitwise_and(frame, frame, mask=green_mask)
    # Display the red result
    
    cv2.imshow('frame', frame)
    cv2.imshow('Red Result', green_result)
    cv2.imshow('Red Mask', green_mask)

    key= cv2.waitKey(1)
    if key == 27:  # ESC key
        cap.release()
        cv2.destroyAllWindows()
        break
  
```
