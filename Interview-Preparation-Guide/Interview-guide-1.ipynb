## **📅 4-Week Data Science Interview Preparation Plan**

---

### **🔹 Week 1 – Core Python, Data Handling & SQL**

**Goal:** Revise the foundations and coding essentials for DS/ML interviews.

| Day | Topics (2-3 hrs weekday / 5-6 hrs weekend)                                                              |
| --- | ------------------------------------------------------------------------------------------------------- |
| Mon | Python Basics (loops, functions, OOP, error handling) – Solve 15 coding problems on LeetCode/HackerRank |
| Tue | Advanced Python (decorators, generators, list comprehension, file handling, regex)                      |
| Wed | NumPy (arrays, indexing, broadcasting, vectorization) + 20 Q\&A                                         |
| Thu | Pandas (dataframes, groupby, merges, time series) + 20 Q\&A                                             |
| Fri | SQL (joins, window functions, subqueries, CTEs) – 15 query problems                                     |
| Sat | Case study: EDA (Exploratory Data Analysis) on Kaggle dataset                                           |
| Sun | Mock test on Python + SQL + Data Wrangling (self-assessment)                                            |

### **🔹 Week 2 – Machine Learning Algorithms & Math Concepts**

**Goal:** Master ML algorithms, intuition, and theory-based interview questions.

| Day | Topics                                                                                             |
| --- | -------------------------------------------------------------------------------------------------- |
| Mon | Supervised Learning (Linear/Logistic Regression, Decision Trees) – theory + practice               |
| Tue | Ensemble Models (Random Forest, XGBoost, LightGBM, Bagging vs Boosting)                            |
| Wed | Unsupervised Learning (K-Means, PCA, DBSCAN, Clustering evaluation metrics)                        |
| Thu | Model Evaluation (accuracy, precision, recall, F1, ROC-AUC, confusion matrix)                      |
| Fri | Overfitting, bias-variance tradeoff, cross-validation, regularization (L1, L2)                     |
| Sat | Math for ML (Probability basics, Bayes theorem, distributions, p-value, t-tests)                   |
| Sun | Mock interview focusing on ML concepts (theory + 2 coding problems on ML implementation in Python) |

| Tue | Hyperparameter tuning (GridSearch, RandomSearch, Bayesian optimization)                                     |

| Thu | Neural Networks & Deep Learning (ANN, CNN, RNN basics)                                                      |
